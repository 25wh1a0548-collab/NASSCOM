```python
import numpy as np
import sympy as sp
x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
    return x**2
h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
import sympy as sp
def g(x):
    return x**3 + 2*x
# 1. Numerical derivative at x = 2 (Finite Difference)
h = 1e-6
x0 = 2
numerical_derivative = (g(x0 + h) - g(x0)) / h
print("Numerical Derivative at x = 2:", numerical_derivative)

# 2. Symbolic derivative
x = sp.Symbol('x')
expr = x**3 + 2*x
symbolic_derivative = sp.diff(expr, x)

print("Symbolic Derivative:", symbolic_derivative)

# 3. Evaluate the symbolic derivative at x = 2
symbolic_value = symbolic_derivative.subs(x, 2)

print("Symbolic Derivative at x = 2:", symbolic_value)
print("Do they match?", abs(numerical_derivative - float(symbolic_value)) < 1e-5)
```

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
import sympy as sp
x, y = sp.symbols('x y')
h2 = x**2 * y + sp.sin(y)
# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)
print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

# 2. Assemble the gradient list
gradient = [dh_dx, dh_dy]
print("\nGradient =", gradient)

# 3. Evaluate at (x=2, y=0)
gradient_at_point = [g.subs({x: 2, y: 0}) for g in gradient]
print("\nGradient at (2, 0) =", gradient_at_point)
```

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)
print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
import sympy as sp
x = sp.Symbol('x')
y = sp.exp(x**2 + 1)
# 1. By hand (Chain Rule)
inner = x**2 + 1
by_hand = sp.exp(inner) * (2 * x)
print("By Hand Derivative:")
print(by_hand)

# 2. With sp.diff
by_sympy = sp.diff(y, x)

print("\nSymPy Derivative:")
print(by_sympy)

# 3. Confirm they match
print("\nDo they match?",
      sp.simplify(by_hand - by_sympy) == 0)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1
z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1
print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np

np.random.seed(42)

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1
Wb = np.random.randn(8, 1) * 0.1

# 1. Forward pass
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb

loss = np.mean((y_hat - Yb) ** 2)
print("Loss before update:", loss)

# 2. Backward pass
n = Xb.shape[0]

dy = (2 / n) * (y_hat - Yb)
dWb = h.T @ dy

dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1

# 3. Gradient Descent Step
lr = 0.05
Wa = Wa - lr * dWa
Wb = Wb - lr * dWb
z1_new = Xb @ Wa
h_new = np.maximum(0, z1_new)
y_hat_new = h_new @ Wb
loss_new = np.mean((y_hat_new - Yb) ** 2)
print("Loss after update:", loss_new)
```

```python
xv = 0.0
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)
    xv = xv - lr * grad
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
import sympy as sp

x, y = sp.symbols('x y')
# 1. Hessian of x**4 + y**2
f = x**4 + y**2
H = sp.hessian(f, (x, y))
print("Hessian of f:")
sp.pprint(H)

# 2. Gradient Descent to minimize (x - 7)**2
xv = 0.0
lr = 0.1
for step in range(20):
    grad = 2 * (xv - 7)    # derivative of (x - 7)^2
    xv = xv - lr * grad
print("\nFinal x:", xv)
```
